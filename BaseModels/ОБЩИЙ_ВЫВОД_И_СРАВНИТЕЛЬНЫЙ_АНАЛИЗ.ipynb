```python
import pandas as pd

comparison_df = pd.DataFrame({
    'Модель': ['Random', 'Most Popular', 'Most Popular (Filtered)',
               'Item-KNN', 'User-KNN', 'SLIM (500)', 'EASE', 'PureSVD (50 factors)'],
    'Категория': ['Бейзлайн', 'Бейзлайн', 'Бейзлайн+', 'Методы схожести', 'Методы схожести',
                  'Линейные методы', 'Линейные методы', 'Матричная факторизация'],
    'NDCG@50 ↑': [0.006, 0.268, 0.438, 0.458, 0.471, 0.460, 0.474, 0.513],
    'Сложность': ['★', '★', '★', '★★', '★★', '★★★', '★', '★★']
})

# Сортируем по NDCG для наглядности
comparison_df = comparison_df.sort_values('NDCG@50 ↑', ascending=False)
```

## **2. Сравнительные результаты всех моделей**

Как видно из сравнительной таблицы, исследование наглядно продемонстрировало значительный прогресс в качестве рекомендаций:

* **Наивные бейзлайны** показали крайне **низкую эффективность**, подтвердив, что простые стратегии не могут уловить персональные предпочтения.

* М**одель на основе популярности** с фильтрацией дала **существенный скачок** в качестве ( **70-кратное улучшение** относительно Random и 2-кратное улучшение относительно базового MostPopular), подчеркнув критическую важность исключения уже известных пользователю айтемов.

* Модели, основанные на схожести (**KNN**) обеспечили дальнейший рост, доказав эффективность принципа *"похожим пользователям нравятся похожие товары"*.

* Матричные методы факторизации (**PureSVD**) продемонстрировали **наилучшие результаты** среди всех классических подходов, эффективно выявляя скрытые латентные факторы.

* Специализированные линейные модели (**EASE**) подтвердили свою конкурентоспособность, показав выдающийся **баланс между высоким качеством и вычислительной эффективностью**.

## **3. Ключевые выводы и перспективы исследования**

Проведенное исследование позволило сформулировать несколько фундаментальных выводов относительно эффективности классических методов коллаборативной фильтрации.

**Во-первых**, выявлена четкая з*ависимость между сложностью модели и качеством* рекомендаций: такие продвинутые подходы, как SVD и EASE, демонстрируют наилучшие метрики, однако требуют значительных вычислительных ресурсов и тщательной настройки гиперпараметров.

**Во-вторых**, подтверждена критическая важность *персонализации* - даже элементарная фильтрация уже просмотренных пользователем элементов приводит к резкому росту качества рекомендаций.

**В-третьих**, установлено, что* линейные модели с аналитическим решением, в частности EASE*, представляют особый практический интерес, поскольку обеспечивают *качество, сопоставимое с SVD*, при существенно меньших вычислительных затратах на обучение и эксплуатацию.

В заключение, исследование подтверждает, что классические методы коллаборативной фильтрации, от KNN до матричной факторизации и современных линейных моделей типа EASE, остаются актуальным и эффективным инструментом, составляющим ядро рекомендательного сервиса. Наилучшие результаты продемонстрировали алгоритмы PureSVD и EASE, которые могут быть рекомендованы в качестве базовых для создания production-решений.

Несмотря на достигнутые показатели, потенциал для улучшения остается значительным. Основными направлениями для дальнейших исследований являются: переход к гибридным моделям, интегрирующим коллаборативную фильтрацию с контентными, временными и социальными признаками, а также внедрение нейросетевых подходов для учета нелинейных взаимодействий и временных последовательностей действий пользователей.
