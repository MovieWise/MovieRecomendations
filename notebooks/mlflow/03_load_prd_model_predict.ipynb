# Load PRD model and predict

Ноутбук находит модель или run с тегом `PRD`, загружает модель из MLflow и делает тестовый predict.

```python
import os
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "mlflow":
    PROJECT_ROOT = PROJECT_ROOT.parents[1]
os.chdir(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT / "src"))
os.environ.setdefault("PYTHONPATH", str(PROJECT_ROOT / "src"))
os.environ.setdefault("MLFLOW_TRACKING_URI", "http://localhost:5000")
os.environ.setdefault("MLFLOW_S3_ENDPOINT_URL", "http://localhost:9000")
os.environ.setdefault("AWS_ACCESS_KEY_ID", "minio")
os.environ.setdefault("AWS_SECRET_ACCESS_KEY", "minio123")
os.environ.setdefault("AWS_DEFAULT_REGION", "us-east-1")
```

```python
import pandas as pd
from movie_recs.training.hybrid_mlflow import PRD_FEATURES, load_prd_model_and_predict

test_path = PROJECT_ROOT / "data" / "processed" / "hybrid_test.parquet"
test_df = pd.read_parquet(test_path)

sample = test_df[PRD_FEATURES].head(5).copy()

for col in ["main_genre", "main_region"]:
    if col in sample.columns:
        sample[col] = sample[col].astype("category")

prediction = load_prd_model_and_predict(sample=sample)
prediction
```
